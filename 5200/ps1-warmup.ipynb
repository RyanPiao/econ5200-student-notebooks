## 1. Draw right-skewed data

`rng.lognormal(mean, sigma, size)` draws positive values with a long right tail. `mean` and `sigma` describe the **log** of the values, so the median of the draw is about `np.exp(mean)`, not `mean`.

**Watch:** run this and read it.

```python
rents = rng.lognormal(mean=np.log(1500), sigma=0.4, size=10_000)

print(f"median {np.median(rents):,.0f}   mean {rents.mean():,.0f}")   # mean > median: right skew
```

## 2. A trimmed mean

`stats.trim_mean(x, 0.1)` drops the top 10% and the bottom 10%, then averages what is left, so a few extreme values barely move it.

**Watch:** run this and read it.

```python
salaries = np.append(rng.normal(60_000, 8_000, size=97), [2e6, 3e6, 5e6])   # three CEOs

print(f"mean    {salaries.mean():>10,.0f}")
print(f"median  {np.median(salaries):>10,.0f}")
print(f"trimmed {stats.trim_mean(salaries, 0.1):>10,.0f}")
```
